```python
import numpy as np
import pandas as pd

# Baseline call counts (integers)
baseline_calls = np.array([
3, 3, 3, 3, 3, 3, 3, 3, 2, 2, 2, 2, 2, 2, 2, 2,
2, 2, 2, 2, 2, 2, 2, 2, 5, 5, 5, 5, 9, 9, 9, 9,
57, 57, 57, 57, 59, 59, 59, 59, 55, 55, 55, 55,
50, 50, 50, 50, 45, 45, 45, 45, 47, 47, 47, 47,
44, 44, 44, 44, 44, 44, 44, 44, 12, 12, 12, 12,
12, 12, 12, 12, 44, 44, 44, 44, 43, 43, 43, 43,
10, 10, 10, 10, 8, 8, 8, 8, 6, 6, 6, 6, 4, 4, 4, 4
])

# Standard deviation of normal noise (adjust as needed)
sigma = 1.0

# Perturb λ with normal noise and clip to avoid negative rates
perturbed_lambda = np.clip(baseline_calls + np.random.normal(0, sigma, size=baseline_calls.shape), a_min=0, a_max=None)

# Simulate Poisson calls using perturbed λ
simulated_calls = np.random.poisson(lam=perturbed_lambda)

# Put into a dataframe
time_index = pd.date_range(start="00:00", periods=96, freq="15min")
df = pd.DataFrame({
    "time": time_index,
    "baseline_calls": baseline_calls,
    "perturbed_lambda": perturbed_lambda,
    "simulated_calls": simulated_calls
})

print(df)
```
